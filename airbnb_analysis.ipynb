# Airbnb Booking Analysis

This notebook analyzes booking patterns, customer behavior, and demand drivers using Airbnb data.

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
bookings = pd.read_csv("../data/bookings.csv")
listings = pd.read_csv("../data/listings.csv")

df = bookings.merge(listings, on="listing_id")
df.head()
```

## Feature Engineering

Creating stay length variable to analyze customer behavior.

```python
df['stay_length'] = (
    pd.to_datetime(df['checkout_date']) - 
    pd.to_datetime(df['checkin_date'])
).dt.days
```

## Stay Length Distribution
    
Analyzing how long customers typically stay.

```python
import matplotlib.pyplot as plt

df['stay_length'].hist()
plt.title("Stay Length Distribution")
plt.xlabel("Days")
plt.ylabel("Frequency")
plt.show()
```

## Price vs Stay Length

Checking if pricing influences stay behavior.

```python
plt.scatter(df['price'], df['stay_length'])
plt.title("Price vs Stay Length")
plt.xlabel("Price")
plt.ylabel("Stay Length")
plt.show()
```

## Key Insights

- Stay duration is short and varied, with no strong concentration
- No clear relationship between price and stay length
- Indicates demand is influenced by factors beyond pricing
